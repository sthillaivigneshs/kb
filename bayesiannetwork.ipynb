```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

m = DiscreteBayesianNetwork([['rain','traffic'],['accident','traffic']])

r = TabularCPD('rain',2,[[0.7],[0.3]])
a = TabularCPD('accident',2,[[0.9],[0.1]])
t = TabularCPD('traffic',2,[[0.95,0.8,0.7,0.1],[0.05,0.2,0.3,0.9]],evidence=['rain','accident'],evidence_card=[2,2])

m.add_cpds(r,a,t)
i = VariableElimination(m)
print(i.query(['traffic'], evidence={'rain':1}))
```
