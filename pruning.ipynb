```python

W=[(0,1,2),(3,4,5),(6,7,8),(0,3,6),(1,4,7),(2,5,8),(0,4,8),(2,4,6)]

def win(b,p):
    return any(all(b[i]==p for i in c) for c in W)

def mini(b,a,al,be):
    if win(b,'X'): return 1
    if win(b,'O'): return -1
    if ' ' not in b: return 0
    for i in range(9):
        if b[i]==' ':
            b[i]='X' if a else 'O'
            v=mini(b,not a,al,be)
            b[i]=' '
            if a: al=max(al,v)
            else: be=min(be,v)
            if be<=al: break
    return al if a else be

def best(b):
    v,m=-2,0
    for i in range(9):
        if b[i]==' ':
            b[i]='X'
            s=mini(b,False,-2,2)
            b[i]=' '
            if s>v: v,m=s,i
    return m

def show(b):
    for i in range(0,9,3):
        print('|'.join(b[i:i+3]))
        if i<6: print('-----')

b=[' ']*9
while True:
    show(b)
    m=int(input("Enter your move (0-8): "))
    if m not in range(9) or b[m]!=' ':
        print("Invalid move"); continue
    b[m]='O'
    if win(b,'O'): show(b); print("Human wins!"); break
    if ' ' not in b: show(b); print("Draw!"); break
    m=best(b)
    b[m]='X'
    print("AI chooses position",m)
    if win(b,'X'): show(b); print("AI wins!"); break
    if ' ' not in b: show(b); print("Draw!"); break
```
