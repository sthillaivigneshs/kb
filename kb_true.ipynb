```python

def neg(x):
    return x[1:] if x.startswith('~') else '~'+x

def resolve(a,b):
    r=[]
    for x in a:
        for y in b:
            if x==neg(y):
                c=set(a+b)-{x,y}
                r.append(list(c))
    return r

def resolution(K,q):
    c=K+[[neg(x) for x in q]]
    while True:
        n=len(c)
        new=set()
        for i in range(n):
            for j in range(i+1,n):
                for r in resolve(c[i],c[j]):
                    if not r: return True
                    new.add(tuple(sorted(r)))
        if new.issubset({tuple(sorted(x)) for x in c}):
            return False
        c += [list(x) for x in new]

KB=[['P','Q'],['~P','R'],['~Q','R']]
print("Query entailed by KB?",resolution(KB,['R']))
```
