```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans, AgglomerativeClustering
import matplotlib.pyplot as plt

x,_=make_blobs(n_samples=150, centers=4, cluster_std=0.60, random_state=0)

kmeans=KMeans(n_clusters=4,random_state=0)
y_keams=kmeans.fit_predict(x)

agg=AgglomerativeClustering(n_clusters=4)
y_agg=agg.fit_predict(x)

fig,(ax1,ax2)=plt.subplots(1,2,figsize=(12,5))
ax1.scatter(x[:,0],x[:,1],c=y_keams,cmap='viridis')
ax1.set_title('K-Means clustering')
ax2.scatter(x[:,0],x[:,1],c=y_agg,cmap='rainbow')
ax2.set_title('Agglomerative clustering')
plt.show()
```

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans, AgglomerativeClustering
import matplotlib.pyplot as plt

x,_=make_blobs(n_samples=150, centers=4, cluster_std=0.60, random_state=0)

kmeans=KMeans(n_clusters=4, random_state=0)
kpred=kmeans.fit_predict(x)
agg=AgglomerativeClustering(n_clusters=4)
apred=agg.fit_predict(x)

fig,(ax1,ax2) = plt.subplots(1,2,figsize=(12,5))
ax1.scatter(x[:,0],x[:,1],c=kpred,cmap='viridis')
ax1.set_title('K-Means clustering')
ax2.scatter(x[:,0],x[:,1],c=apred,cmap='rainbow')
ax2.set_title('Agglomerative clustering')
plt.show()
```
